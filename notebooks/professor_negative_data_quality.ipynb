# Professor negative audio: data-quality assessment

## TL;DR

The ten recordings are technically readable and contain useful traffic/forest hard negatives, but they are **not ready for training**. Use them conditionally as a held-out private evaluation set after listening to the flagged timestamps. The central risk is a 48 kHz negative vs 16 kHz professor-positive sample-rate confound and strong model sensitivity to resampling.

## Context and methods

This notebook performs a non-destructive audit. It inventories WAV metadata, checks level/clipping/DC/dropout proxies, extracts the frozen `spectral_mfcc_v1` feature contract at native 48 kHz and after in-memory 16 kHz resampling, and evaluates the two frozen models without retuning thresholds. Recording-level bootstrap intervals treat each WAV—not each overlapping window—as the independent unit. No clean reference is available, so true SNR cannot be estimated. Automated model scores flag review targets; they do not validate labels.

```python
from pathlib import Path
import csv
import json
import sys

repo_root = Path.cwd()
if not (repo_root / 'scripts').is_dir():
    repo_root = repo_root.parent
sys.path.insert(0, str(repo_root / 'scripts'))
from analyze_private_negative_audio import run_analysis

out_dir = repo_root / 'artifacts/analysis/private_negative_quality'
summary = run_analysis(
    negative_root=repo_root / 'data/raw/private/Recordings',
    positive_wav=repo_root / 'data/raw/private/professor_real_world/SMU08662_20260608_135349.wav',
    feature_bundle=repo_root / 'artifacts/features/v2_spectral_mfcc_features.npz',
    scaler_path=repo_root / 'artifacts/features/v2_spectral_mfcc_scaler.json',
    model_paths={
        'tiny_mlp_es': repo_root / 'artifacts/models/v2b_mlp/tiny_mlp_es.joblib',
        'random_forest': repo_root / 'artifacts/models/v2a_sweep/models/random_forest.joblib',
    },
    out_dir=out_dir,
    progress_every=250,
)
```

## Data inventory and signal integrity

```python
print(json.dumps({
    'scope': summary['scope'],
    'audio_format': summary['audio_format'],
    'signal_flags': summary['signal_quality']['flags'],
    'duplicates': summary['duplicate_checks'],
}, indent=2))

with (out_dir / 'file_profile.csv').open(newline='') as handle:
    profiles = list(csv.DictReader(handle))
print('\nfile | env | duration_s | median_dBFS | level_spread_dB')
for row in profiles:
    print(f"{row['file']} | {row['environment']} | {float(row['duration_sec']):.1f} | "
          f"{float(row['one_sec_rms_dbfs_p50']):.1f} | {float(row['one_sec_level_spread_db']):.1f}")
```

## Frozen-model stress test

False-positive rates below are diagnostic, not final benchmark claims: timestamps overlap and labels have not been manually audited. The recording-level macro estimate and bootstrap interval are the more defensible uncertainty summary.

```python
evaluation = summary['frozen_model_evaluation']
print('model | public_negative_FPR | native48k_FPR | resampled16k_FPR | recording_macro_16k [95% CI]')
for model in ('tiny_mlp_es', 'random_forest'):
    public = evaluation['public_test_negative_reference'][model]['false_positive_rate']
    native = evaluation['new_negative']['native_48k']['models'][model]['false_positive_rate']
    sampled = evaluation['new_negative']['resampled_16k']['models'][model]['false_positive_rate']
    recording = evaluation['new_negative']['resampled_16k']['recording_level_models'][model]
    print(f"{model} | {public:.3%} | {native:.3%} | {sampled:.3%} | "
          f"{recording['macro_false_positive_rate']:.3%} "
          f"[{recording['bootstrap_95pct_ci_lower']:.3%}, {recording['bootstrap_95pct_ci_upper']:.3%}]")
print('\nResampling sensitivity:')
print(json.dumps(evaluation['resampling_sensitivity'], indent=2))
```

## Domain shift and label-review targets

```python
print(json.dumps(summary['domain_shift']['by_mode'], indent=2))
print(f"\nConsensus review targets: {summary['label_review']['consensus_suspect_windows']}")
with (out_dir / 'suspect_windows.csv').open(newline='') as handle:
    suspects = list(csv.DictReader(handle))
for row in suspects:
    print(f"{row['file']} {float(row['window_start_sec']):.1f}-{float(row['window_end_sec']):.1f}s "
          f"MLP={float(row['tiny_mlp_es_score']):.3f} RF={float(row['random_forest_score']):.3f}")
```

## Takeaways

1. Keep the source WAVs unchanged; environmental noise is the target negative domain, not a defect to remove.
2. Manually listen to and annotate the consensus suspect windows before accepting the negative labels.
3. Register the accepted files as a held-out private hard-negative evaluation set only. Do not tune thresholds or select models on it.
4. Enforce one sample-rate policy across public training, professor positive, professor negative, and inference, then rebuild features and retrain before considering these recordings for training.
5. Collect more independent days/devices/environments; ten recordings from two sessions do not establish deployment generalization.
